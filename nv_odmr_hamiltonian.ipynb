# NV Hamiltonian: appendix calculations

Exact spin-1 diagonalization, weak-transverse-field approximation and analytic
cubic roots. Fields are expressed in the **local NV frame**, independently of
the experimental image coordinates. `E=5 MHz` is illustrative, not a measured
strain calibration. The two shift plots retain `gamma_e=28.0 GHz/T` from the
original figures. The model-error table uses `28.024 GHz/T`, as in the field
conversion of the revised manuscript.


```python
%matplotlib inline
from pathlib import Path
ROOT = Path.cwd()
assert (ROOT / "helpers.py").is_file(), "Launch from the repository root."
OUT = ROOT / "outputs" / "hamiltonian"
OUT.mkdir(parents=True, exist_ok=True)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from nv_hamiltonian_helpers import (H_over_h, odmr_frequencies,
                                    analytic_eigenvalues, odmr_frequencies_analytic)
D, E, gamma_e = 2.87, 0.005, 28.0
Bperp_mT = np.linspace(0, 20, 100)
Bperp = Bperp_mT * 1e-3
```

```python
Bz = 0.015
figure_name = 'shift_Db.png'
nu_minus_approx = D - np.sqrt((gamma_e * Bz)**2 + E**2)
nu_plus_approx  = D + np.sqrt((gamma_e * Bz)**2 + E**2)


# ============================================================
# Compute exact curves for different transverse-field directions
# ============================================================

cases = {
    r"$B_x=B_\perp,\ B_y=0$": lambda B: (B, 0.0),
    #r"$B_x=0,\ B_y=B_\perp$": lambda B: (0.0, B),
    r"$B_x=B_y=B_\perp/\sqrt{2}$": lambda B: (B / np.sqrt(2), B / np.sqrt(2)),
    #r"$B_x=B_y=B_\perp/\sqrt{2}$": lambda B: (B, B),
}

results = {}

for label, field_components in cases.items():
    nu_minus = []
    nu_plus = []

    for B in Bperp:
        Bx, By = field_components(B)
        n1, n2 = odmr_frequencies(Bx, By, Bz)
        nu_minus.append(n1)
        nu_plus.append(n2)

    results[label] = {
        "nu_minus": np.array(nu_minus),
        "nu_plus": np.array(nu_plus),
    }


# ============================================================
# Plot: nu_minus and nu_plus vs transverse field
# ============================================================

plt.figure(figsize=(8, 5))

#for label, data in results.items():
#    plt.plot(Bperp_mT, data["nu_plus"], label=rf"$\nu_+$ exact, {label}", linestyle="--")
#    plt.plot(Bperp_mT, data["nu_minus"], label=rf"$\nu_-$ exact, {label}")
tmp=0
for label, data in results.items():
    if tmp == 1:
        ls="--" 
        col="seagreen"
        mk="+"
        lw=2.5
    else:
        ls="-"
        col="lime"
        mk="."
        lw=2.5
    plt.plot(Bperp_mT, data["nu_plus"], label=rf"$\nu_\pm$ exact, {label}", linestyle=ls, color=col, linewidth=lw)
    plt.plot(Bperp_mT, data["nu_minus"], linestyle=ls, color=col, linewidth=lw)

    if tmp==1:
        plt.plot(Bperp_mT, (data["nu_plus"]+data["nu_minus"])/2.0, label=rf"$(\nu_+ + \nu_-)/2$", linestyle="--", color="black", linewidth=lw)
    tmp=tmp+1



# Symmetry line around D
plt.axhline(D, linestyle="-", linewidth=2.5, label=r"symmetry line $D$", color="black")

# Low-transverse-field approximation at B_perp = 0
#plt.axhline(nu_minus_approx, linestyle=":", linewidth=2, label=r"$\nu_- \simeq D-\sqrt{(\gamma_e B_z)^2+E^2}$")
plt.axhline(nu_minus_approx, linestyle="-", linewidth=2, color="mediumblue")
plt.axhline(nu_plus_approx, linestyle="-", linewidth=2, color="mediumblue", label=r"$\nu_\pm \simeq D\pm\sqrt{(\gamma_e B_z)^2+E^2}$",)

plt.xlabel(r"$B_\perp$ [mT]")
plt.ylabel("ODMR frequencies [GHz]")
plt.title(rf"Exact ODMR transition frequencies for $B_z={Bz*1e3:.0f}$ mT, $E={E*1e3:.1f}$ MHz")

plt.grid(True, alpha=0.3)
plt.legend(fontsize=13, loc="lower center", bbox_to_anchor=(0.22, 0.52))
plt.tight_layout()
plt.savefig(OUT / figure_name, dpi=200, bbox_inches="tight")
plt.show()


```

```python
# Systematic model effects, not experimental confidence intervals.
rows = []
for perp_mT in (0, 5, 10):
    low, high = odmr_frequencies(perp_mT*1e-3, 0, 5e-3, gamma_e=28.024)
    rows.append(dict(Bpar_mT=5, Bperp_mT=perp_mT,
                     estimated_lower_mT=abs(low-D)/28.024*1e3,
                     estimated_upper_mT=abs(high-D)/28.024*1e3))
model_error = pd.DataFrame(rows)
display(model_error)
model_error.to_csv(OUT / 'single_resonance_model_error.csv', index=False)
```

```python
# Check cubic roots against independent numerical diagonalization.
rows = []
for Bz in (5e-3, 15e-3):
    for perp in Bperp:
        for bx, by in ((perp, 0), (0, perp), (perp/np.sqrt(2), perp/np.sqrt(2))):
            numerical = np.linalg.eigvalsh(H_over_h(bx, by, Bz))
            analytical = analytic_eigenvalues(bx, by, Bz)
            rows.append(np.max(np.abs(numerical-analytical)))
max_eigenvalue_error_GHz = float(np.max(rows))
assert max_eigenvalue_error_GHz < 1e-12
print('Maximum eigenvalue discrepancy [GHz]:', max_eigenvalue_error_GHz)
(OUT / 'analytic_roots_check.txt').write_text(str(max_eigenvalue_error_GHz)+'\n')
```
